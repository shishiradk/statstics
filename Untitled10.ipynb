```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#define our dataset
dataset=[11,10,12,14,12,15,14,13,15,102,12,14,11,12,13,14,16]
```

```python
plt.hist(dataset)
```

```python
outliers=[]

def detect_outliers(data):
  threshold=3
  mean=np.mean(data)
  std=np.std(data)

  for i in data:
    z_score=(i-mean)/std
    if np.abs(z_score)>threshold:
      outliers.append(i)

      return outliers
```

```python
detect_outliers(dataset)
```

```python
dataset = sorted(dataset)
dataset
```

```python
q1,q3=np.percentile(dataset,[25,75])
```

```python
 print(q1,q3)
```

```python
iqr=q3-q1
print(iqr)
```

```python
lower_fence=q1-(1.5*iqr)
higher_fence=q3+(1.5*iqr)
```

```python
print(lower_fence,higher_fence)
```
